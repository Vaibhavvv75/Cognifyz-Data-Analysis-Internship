```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(r"C:\Users\DX002\Downloads\Dataset .csv")

df.head()
```

```python
df[["Restaurant Name", "City", "Longitude", "Latitude"]].head(10)
```

```python
df[["Longitude", "Latitude"]].isnull().sum()
```

```python
plt.figure(figsize=(10, 7))

plt.scatter(
    df["Longitude"],
    df["Latitude"],
    alpha=0.5
)

plt.title("Geographic Distribution of Restaurants")
plt.xlabel("Longitude")
plt.ylabel("Latitude")

plt.tight_layout()
plt.show()
```

```python
city_counts = df["City"].value_counts().head(10)

city_counts
```

```python
plt.figure(figsize=(10, 6))

plt.barh(
    city_counts.index[::-1],
    city_counts.values[::-1]
)

plt.title("Top 10 Cities by Number of Restaurants")
plt.xlabel("Number of Restaurants")
plt.ylabel("City")

plt.tight_layout()
plt.show()
```

```python
city_percentage = (city_counts / len(df)) * 100

city_summary = pd.DataFrame({
    "City": city_counts.index,
    "Restaurant Count": city_counts.values,
    "Percentage": city_percentage.values.round(2)
})

city_summary
```

## Observation

- New Delhi has the highest number of restaurants in the dataset, with 5,473 restaurants (57.30%).
- Gurgaon has 1,118 restaurants (11.71%), while Noida has 1,080 restaurants (11.31%).
- Faridabad has 251 restaurants (2.63%).
- The restaurant distribution is highly concentrated in New Delhi, Gurgaon, and Noida.
- The geographic scatter plot also shows a strong concentration of restaurant locations within the major restaurant-dense regions.
- Overall, the analysis indicates that restaurants are not evenly distributed geographically and are concentrated in a few major cities.
