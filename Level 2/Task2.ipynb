```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(r"C:\Users\DX002\Downloads\Dataset .csv")

df.head()
```

```python
cuisine_combinations = df["Cuisines"].value_counts()

cuisine_combinations.head(10)
```

```python
top_10_combinations = cuisine_combinations.head(10)

plt.figure(figsize=(10, 6))

plt.barh(
    top_10_combinations.index[::-1],
    top_10_combinations.values[::-1]
)

plt.title("Top 10 Most Common Cuisine Combinations")
plt.xlabel("Number of Restaurants")
plt.ylabel("Cuisine Combination")

plt.tight_layout()
plt.show()
```

```python
combination_ratings = df.groupby("Cuisines")["Aggregate rating"].agg(
    ["count", "mean"]
)

combination_ratings = combination_ratings[
    combination_ratings["count"] >= 10
]

top_rated_combinations = combination_ratings.sort_values(
    "mean",
    ascending=False
).head(10)

top_rated_combinations
```
