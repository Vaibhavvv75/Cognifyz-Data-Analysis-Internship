```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(r"C:\Users\DX002\Downloads\Dataset .csv")

df[["Longitude", "Latitude"]].head()
```

```python
plt.figure(figsize=(10, 7))

plt.scatter(
    df["Longitude"],
    df["Latitude"],
    alpha=0.5
)

plt.title("Geographic Distribution of Restaurants")
plt.xlabel("Longitude")
plt.ylabel("Latitude")

plt.tight_layout()
plt.show()
```

```python
chain_counts = df["Restaurant Name"].value_counts()

chain_counts.head(10)
```

```python
top_10_chains = chain_counts.head(10)

plt.figure(figsize=(10, 6))

plt.barh(
    top_10_chains.index[::-1],
    top_10_chains.values[::-1]
)

plt.title("Top 10 Most Common Restaurant Chains")
plt.xlabel("Number of Restaurants")
plt.ylabel("Restaurant Chain")

plt.tight_layout()
plt.show()
```

```python
chain_analysis = df.groupby("Restaurant Name").agg(
    Restaurant_Count=("Restaurant ID", "count"),
    Average_Rating=("Aggregate rating", "mean"),
    Average_Votes=("Votes", "mean")
)

chain_analysis = chain_analysis[
    chain_analysis["Restaurant_Count"] >= 5
]

chain_analysis = chain_analysis.sort_values(
    "Restaurant_Count",
    ascending=False
)

chain_analysis.head(10)
```

```python
chain_analysis = df.groupby("Restaurant Name").agg(
    Restaurant_Count=("Restaurant ID", "count"),
    Average_Rating=("Aggregate rating", "mean"),
    Average_Votes=("Votes", "mean")
)

chain_analysis = chain_analysis[
    chain_analysis["Restaurant_Count"] >= 5
]

chain_analysis = chain_analysis.sort_values(
    "Restaurant_Count",
    ascending=False
)

chain_analysis.head(10)
```

## Observation

- Cafe Coffee Day has the highest number of restaurants in the analyzed chains, with 83 outlets.
- Domino's Pizza and Subway are also among the most widely present restaurant chains.
- Barbeque Nation has the highest average rating among the top analyzed chains, with an average rating of 4.35.
- Barbeque Nation also has the highest average votes, with approximately 1,082 votes per restaurant, indicating strong popularity and customer engagement.
- McDonald's and Pizza Hut also have relatively high average ratings of 3.34 and 3.32 respectively.
- The analysis shows that restaurant chains with fewer outlets can still have higher ratings and customer engagement.
