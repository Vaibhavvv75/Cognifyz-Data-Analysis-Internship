```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\DX002\Downloads\Dataset .csv")
```

```python
df["Aggregate rating"].value_counts().sort_index()
```

```python
plt.figure(figsize=(10, 6))

plt.bar(
    df["Aggregate rating"].value_counts().sort_index().index,
    df["Aggregate rating"].value_counts().sort_index().values
)

plt.title("Restaurant Rating Distribution")
plt.xlabel("Aggregate Rating")
plt.ylabel("Number of Restaurants")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
average_rating = df["Aggregate rating"].mean()

print("Average Restaurant Rating:", round(average_rating, 2))
```

## Observation

- The average restaurant rating in the dataset is **2.67**.
- The rating distribution ranges from **0.0 to 4.9**.
- A large number of restaurants have a rating of **0.0**, with **2,148 restaurants**.
- Among the rated restaurants, ratings are mainly concentrated around the **3.0–4.0** range.
- The rating distribution helps understand the overall customer rating pattern across restaurants.
