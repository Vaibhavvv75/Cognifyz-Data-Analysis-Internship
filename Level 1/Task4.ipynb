```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\DX002\Downloads\Dataset .csv")
```

```python
delivery_counts = df["Has Online delivery"].value_counts()

delivery_counts
```

```python
delivery_percentage = (delivery_counts / len(df)) * 100

delivery_percentage.round(2)
```

```python
delivery_summary = pd.DataFrame({
    "Online Delivery": delivery_counts.index,
    "Restaurant Count": delivery_counts.values,
    "Percentage": delivery_percentage.values.round(2)
})

delivery_summary
```

```python
delivery_rating = df.groupby("Has Online delivery")["Aggregate rating"].mean()

delivery_rating.round(2)
```

```python
plt.figure(figsize=(7, 5))

plt.bar(
    delivery_rating.index,
    delivery_rating.values
)

plt.title("Average Rating: Online Delivery vs No Online Delivery")
plt.xlabel("Online Delivery")
plt.ylabel("Average Rating")

plt.show()
```

## Observation

- Restaurants offering online delivery have an average rating of 3.25.
- Restaurants without online delivery have an average rating of 2.47.
- The average rating is higher for restaurants that offer online delivery.
- The delivery availability and rating comparison provide insights into the relationship between online delivery and restaurant ratings.
