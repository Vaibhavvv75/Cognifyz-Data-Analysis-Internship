```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\DX002\Downloads\Dataset .csv")
```

```python
df.head()
```

```python
city_counts = df["City"].value_counts()

city_counts.head(10)
```

```python
city_avg_rating = df.groupby("City")["Aggregate rating"].mean()

city_avg_rating.head(10)
```

```python
highest_avg_city = city_avg_rating.idxmax()
highest_avg_rating = city_avg_rating.max()

print("City with highest average rating:", highest_avg_city)
print("Highest average rating:", round(highest_avg_rating, 2))
```

```python
summary = pd.DataFrame({
    "Analysis": [
        "City with highest number of restaurants",
        "Number of restaurants",
        "City with highest average rating",
        "Highest average rating"
    ],
    "Result": [
        city_counts.idxmax(),
        city_counts.max(),
        highest_avg_city,
        round(highest_avg_rating, 2)
    ]
})

summary
```

```python
top_10_rating_cities = city_avg_rating.sort_values(ascending=False).head(10)

plt.figure(figsize=(10, 6))
plt.bar(top_10_rating_cities.index, top_10_rating_cities.values)

plt.title("Top 10 Cities by Average Restaurant Rating")
plt.xlabel("City")
plt.ylabel("Average Rating")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Observation

- New Delhi has the highest number of restaurants in the dataset.
- City-wise average ratings vary across different cities.
- The city with the highest average restaurant rating was identified using the mean of the Aggregate rating column.
- The top 10 cities by average rating are visualized in the bar chart above.
