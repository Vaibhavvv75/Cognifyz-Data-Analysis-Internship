```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\DX002\Downloads\Dataset .csv")
```

```python
df["Price range"].value_counts().sort_index()
```

```python
price_counts = df["Price range"].value_counts().sort_index()

price_percentage = (price_counts / len(df)) * 100

price_percentage
```

```python
price_distribution = pd.DataFrame({
    "Price Range": price_counts.index,
    "Restaurant Count": price_counts.values,
    "Percentage": price_percentage.values.round(2)
})

price_distribution
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    price_distribution["Price Range"],
    price_distribution["Restaurant Count"]
)

plt.title("Price Range Distribution")
plt.xlabel("Price Range")
plt.ylabel("Number of Restaurants")

plt.show()
```

## Observation

- Price Range 1 has the highest number of restaurants, with 4,444 restaurants (46.53%).
- Price Range 2 has 3,113 restaurants (32.59%).
- Price Range 3 has 1,408 restaurants (14.74%).
- Price Range 4 has the lowest number of restaurants, with 586 restaurants (6.14%).
- The majority of restaurants fall within Price Range 1 and Price Range 2.
